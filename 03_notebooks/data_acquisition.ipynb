```python
def extract_zip_files(folder_name):
    for zip_path in folder_name.glob("*.zip"):
        print("Extracting: ", zip_path.name)

        with zipfile.ZipFile(zip_path, "r") as zip_ref:
            zip_ref.extractall(EXTRACTED_FILES)

    print("Completed: ", folder_name.name)
```

```python
combined_file = EXTRACTED_FILES / "combined_flights_2024_2025.csv"

first_file = True

for file in csv_files:
    print("Processing:", file.name)

    for chunk in pd.read_csv(file, chunksize=100_000, low_memory=False):
        chunk.to_csv(
            combined_file,
            mode="w" if first_file else "a",
            index=False,
            header=first_file
        )
        first_file = False

print("Combined CSV created:", combined_file)
```

```python
total_rows = 0

for chunk in pd.read_csv(
    combined_file,
    chunksize=100_000,
    low_memory=False
):
    total_rows += len(chunk)

print("Total rows:", total_rows)
```

```python
missing_values = None

for chunk in pd.read_csv(
    combined_file,
    chunksize=100_000,
    low_memory=False
):
    current_missing = chunk.isna().sum()

    if missing_values is None:
        missing_values = current_missing
    else:
        missing_values = missing_values.add(
            current_missing,
            fill_value=0
        )

print("Overall Missing Values:")
print(missing_values)
```

```python
missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (missing_values / total_rows) * 100
})

missing_table = missing_table.sort_values(
    "Missing Percentage",
    ascending=False
)

missing_table
```

```python
empty_columns = missing_table[
    missing_table["Missing Values"] == total_rows
]

print("Completely empty columns:", len(empty_columns))
display(empty_columns)
```

```python
original_total_rows = 0

for file in csv_files:
    rows = 0

    for chunk in pd.read_csv(file, chunksize=100_000, low_memory=False):
        rows += len(chunk)

    print(file.name, "->", rows, "rows")
    original_total_rows += rows

print("\nTotal rows in original CSV files:", original_total_rows)
```

```python
combined_total_rows = 0

for chunk in pd.read_csv(combined_file, chunksize=100_000, low_memory=False):
    combined_total_rows += len(chunk)

print("Original CSV total rows :", original_total_rows)
print("Combined CSV total rows :", combined_total_rows)

if original_total_rows == combined_total_rows:
    print("✅ Row counts match. No rows were lost during combining.")
else:
    print("❌ Row counts do not match.")
    print("Difference:", original_total_rows - combined_total_rows)
```

```python
original_missing = None

for file in csv_files:
    print("Checking:", file.name)

    for chunk in pd.read_csv(file, chunksize=100_000, low_memory=False):
        current_missing = chunk.isna().sum()

        if original_missing is None:
            original_missing = current_missing
        else:
            original_missing = original_missing.add(
                current_missing,
                fill_value=0
            )

print("Original files checked.")
```

```python
combined_missing = None

for chunk in pd.read_csv(combined_file, chunksize=100_000, low_memory=False):
    current_missing = chunk.isna().sum()

    if combined_missing is None:
        combined_missing = current_missing
    else:
        combined_missing = combined_missing.add(
            current_missing,
            fill_value=0
        )

print("Combined file checked.")
```

```python
missing_comparison = pd.DataFrame({
    "Original Missing": original_missing,
    "Combined Missing": combined_missing
})

missing_comparison["Difference"] = (
    missing_comparison["Combined Missing"]
    - missing_comparison["Original Missing"]
)

display(missing_comparison)
```
